## 3. Process several records

````python
def double(number):
    return number * 2

print(double(5))
````

## 6. Store results instead of only printing them

We can start with an empty list and add matching IDs using `.append()`.


````python
selected_ids = []

for post in posts:
    if post["likes"] >= 10:
        selected_ids.append(post["id"])

print(selected_ids)
````

## 8. Reading a loop error

Look at this example:

```python
for post in posts
    print(post["id"])
```

What is missing?

The answer is the colon `:` after `posts`.

Another common problem is indentation:

```python
for post in posts:
print(post["id"])
```

The line inside the loop must be indented.
