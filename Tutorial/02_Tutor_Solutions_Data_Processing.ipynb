## 1. `for` loops

A `for` loop repeats an operation for each item in a collection.


````python
likes = [5, 12, 8, 20]

for value in likes:
    print(value)
````

### Mini-task 1

Change the loop so that it prints each value multiplied by `2`.

## 2. Combine a loop with a condition

We already know how to ask a yes/no question using `if`.

Now we can ask that question for every value in a list.


````python
likes = [5, 12, 8, 20]

for value in likes:
    if value >= 10:
        print(value)
````

### Mini-task 2

Using the same list:

1. print only values below `10`,
2. then change the condition so that only values of at least `15` are printed.


````python
likes = [5, 12, 8, 20]

for value in likes:
    if value < 10:
        print(value)

print("At least 15:")

for value in likes:
    if value >= 15:
        print(value)
````

## 3. Process several records

Each social-media post is represented by a dictionary, and all posts are stored in a list.


````python
posts = [
    {"id": "P01", "text": "Bus service was good today", "likes": 8},
    {"id": "P02", "text": "The tram was delayed again this morning", "likes": 15},
    {"id": "P03", "text": "More cycling lanes would improve the city", "likes": 21},
    {"id": "P04", "text": "Walking to campus is usually convenient", "likes": 6}
]

print(posts[0])
````

A loop lets us process every post:


````python
for post in posts:
    print(post["id"], post["likes"])
````

### Mini-task 3

Print the ID of every post with **at least 10 likes**.

Expected IDs:

```text
P02
P03
```


````python
for post in posts:
    if post["likes"] >= 10:
        print(post["id"])
````

### Mini-task 4

Print both the ID and the number of likes for posts with at least `10` likes.

Expected:

```text
P02 15
P03 21
```


````python
for post in posts:
    if post["likes"] >= 10:
        print(post["id"], post["likes"])
````

## 4. Functions

A function lets us give a reusable name to a calculation.


````python
def double(number):
    return number * 2

print(double(5))
double(5)
````

Now we create a function that counts the number of words in a text.


````python
def count_words(text):
    return len(text.split())

print(count_words("Python is useful"))
````

### Mini-task 5

Use `count_words()` to count the words in:

```text
The tram was delayed again this morning
```

Expected result: `7`


````python
text = "The tram was delayed again this morning"
print(count_words(text))
````

## 5. Combine functions, loops and conditions

Now we can use our function while processing several posts.


````python
for post in posts:
    words = count_words(post["text"])
    print(post["id"], words)
````

### Mini-task 6

Print only the posts that contain **at least 7 words**.

For each matching post, print its ID and word count.

Expected:

```text
P02 7
P03 7
```


````python
for post in posts:
    words = count_words(post["text"])
    if words >= 7:
        print(post["id"], words)
````

## 6. Store results instead of only printing them

We can start with an empty list and add matching IDs using `.append()`.


````python
selected_ids = []

for post in posts:
    if post["likes"] >= 10:
        selected_ids.append(post["id"])

print(selected_ids)
````

### Mini-task 7

Create a list called `long_post_ids`.

Add the ID of every post containing at least `7` words.

Expected:

```python
['P02', 'P03']
```


````python
long_post_ids = []

for post in posts:
    if count_words(post["text"]) >= 7:
        long_post_ids.append(post["id"])

print(long_post_ids)
````

## 7. Combined challenge

Create a list called `selected_posts`.

Add the ID of every post that:

- has at least `10` likes, **and**
- contains at least `7` words.

Expected:

```python
['P02', 'P03']
```

Hint: use `and`.


````python
selected_posts = []

for post in posts:
    if post["likes"] >= 10 and count_words(post["text"]) >= 7:
        selected_posts.append(post["id"])

print(selected_posts)
````

## 8. Reading a loop error

Look at this example:

```python
for post in posts
    print(post["id"])
```

What is missing?

The answer is the colon `:` after `posts`.

Another common problem is indentation:

```python
for post in posts:
print(post["id"])
```

The line inside the loop must be indented.

## Common errors

- missing `:` after `for` or `if`
- wrong indentation
- using `posts["id"]` instead of `post["id"]`
- using `post.likes` instead of `post["likes"]`
- forgetting `return`
- calling `count_words(post)` instead of `count_words(post["text"])`
- forgetting to create an empty list before `.append()`
- using `or` instead of `and`
