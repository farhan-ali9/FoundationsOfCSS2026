```python
## Task 2: Numbers
#A student receives these quiz scores:
score1 = 62
score2 = 75
score3 = 83
quiz_scores = [score1, score2, score3]
#Calculate and print the average.
average = (score1 + score2 + score3) / 3
print(f"The average score is: {average}")

average_score = sum(quiz_scores) / len(quiz_scores)
print(f"The average score is: {average_score}")
```

```python
## Task 3: Lists
#Create:
scores = [62, 75, 83, 49]


#1. Print the first score.
print(f"The first score is: {scores[0]}")
#2. Print the last score.
print(f"The last score is: {scores[-1]}")
#3. Add `91`.
scores.append(91)
#4. Print the number of scores.
print(f"The number of scores is: {len(scores)}")
#5. Print the average.
print(f"The average score is: {sum(scores) / len(scores)}")
```

```python
## Task 5: Comparisons
likes = 12
#Write expressions that check:
#1. whether likes are at least 10,
print(likes >= 10)
#2. whether likes equal exactly 12,
print(likes == 12)  
#3. whether likes are below 5
print(likes < 5)
```

```python
## Task 6: Conditions
#Write an `if` / `else` statement:
#if a score is at least 50, print `Pass` else print `Not yet passed`

score = 49
if score >= 50:
    print("Pass")
else:
    print("Not yet passed")

print("second approach")
for score in [49, 50, 80]:
    if score >= 50:
        print("Pass")
    else:
        print("Not yet passed")
```

```python
## Task 7: Combined task
student = {
    "name": "Alex",
    "scores": [55, 70, 40]
}

# 1. calculate the average,
average = sum(student["scores"]) / len(student["scores"])
# 2. print the name and average,
print(f"Name: {student['name']}, Average: {average}")       
# 3. print `Pass` if the average is at least 50, otherwise `Not yet passed`.
if average >= 50:
    print("Pass")
else:
    print("Not yet passed")

for score in student["scores"]:
    if score >= 50:
        print(f"Score: {score} - Pass")
    else:
        print(f"Score: {score} - Not yet passed")
```
