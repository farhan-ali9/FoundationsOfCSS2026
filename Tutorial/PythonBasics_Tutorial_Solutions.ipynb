## 5. Comparisons: ask yes/no questions

A comparison produces either `True` or `False`.


````python
likes = 12

print(likes >= 10)
print(likes == 12)
print(likes < 5)
````

## 6. Conditions: let Python make a decision

An `if` statement runs code when its condition is true.


````python
likes = 12

if likes >= 10:
    print("This post reached the threshold")
else:
    print("This post did not reach the threshold")
````

## 7. Small combined challenge

You now have everything needed for this task.

A student has:
- a name,
- a list of three quiz scores.

Calculate the average score and print whether the student has passed.

For this exercise, a score of **50 or more** is a pass.


````python
student = {
    "name": "Maya",
    "scores": [60, 45, 75]
}

average = sum(student["scores"]) / len(student["scores"])
print(f'{student["name"]} has an average score of {average:.1f}')

# Or print(f'{student["name"]} has an average score of {average}')

if average >= 50:
    print("Pass")
else:
    print("Not yet passed")
````

## 8. Reading errors instead of fearing them

Errors are part of programming.

When something fails:
1. look at the **last line** of the error,
2. identify the error type,
3. look at the line Python points to,
4. fix one thing and run again.

Try to identify the problem before fixing this:


````python
# Uncomment these two lines to test the error:
# likes = 12
# print("Likes: " + likes)
````

The problem is that Python cannot directly join a string and an integer using `+`.

Two fixes are:

```python
print("Likes: " + str(likes))
print(f"Likes: {likes}")
```

## Common beginner errors
- forgetting quotation marks around text
- using `=` inside a comparison
- missing `:` after `if` or `else`
- wrong indentation
- using list index `1` for the first item
- misspelling a dictionary key
- trying to join strings and integers with `+`

2. "What does the last line of the error say?"
